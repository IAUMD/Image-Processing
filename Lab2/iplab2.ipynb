## Procedural Steps: Sampling, Quantization, Arithmetic, and Logical Operations

```python
# Step 1: Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from skimage import data, color
```

```python
# Step 2: Define helper functions for sampling, quantization, and plotting
def sample_image(image, factor):
    """
    Downsamples the image by the given factor.
    """
    height, width = image.shape[:2]
    sampled_image = cv2.resize(image, (width // factor, height // factor), interpolation=cv2.INTER_NEAREST)
    return sampled_image

def quantize_image(image, levels):
    """
    Reduces the number of grayscale levels in the image.
    """
    quantized_image = np.floor(image / (256 // levels)) * (256 // levels)
    quantized_image = quantized_image.astype(np.uint8)
    return quantized_image

def plot_images(original, sampled, quantized):
    """
    Plots the original, sampled, and quantized images side by side.
    """
    plt.figure(figsize=(12, 4))
    
    plt.subplot(1, 3, 1)
    plt.imshow(original, cmap='gray')
    plt.title('Original Image')
    plt.axis('off')
    
    plt.subplot(1, 3, 2)
    plt.imshow(sampled, cmap='gray')
    plt.title('Sampled Image')
    plt.axis('off')
    
    plt.subplot(1, 3, 3)
    plt.imshow(quantized, cmap='gray')
    plt.title('Quantized Image')
    plt.axis('off')
    plt.show()
```

```python
# Step 3: Run baseline sampling and quantization using skimage dataset image
sampling_factor = 14
quantization_levels = 9

# Load sample gray image from skimage dataset
original_image = data.camera()

sampled_image = sample_image(original_image, sampling_factor)
quantized_image = quantize_image(original_image, quantization_levels)
plot_images(original_image, sampled_image, quantized_image)
```

## Task 1: Varying Sampling and Quantization Parameters
**Objective:** Experiment with different sampling factors and quantization levels to observe spatial degradation (pixelation) and gray-level degradation (false contouring).

```python
# Test 1: Moderate Sampling & Quantization
sampled_mod = sample_image(original_image, factor=4)
quantized_mod = quantize_image(original_image, levels=16)
plot_images(original_image, sampled_mod, quantized_mod)

# Test 2: Severe Sampling & High-Level Quantization Reduction
sampled_severe = sample_image(original_image, factor=16)
quantized_high = quantize_image(original_image, levels=2)
plot_images(original_image, sampled_severe, quantized_high)
```

**Observation:**  
* Increasing the **sampling factor** reduces spatial resolution, causing pixelation and blockiness.
* Decreasing the **quantization levels** restricts continuous tone variations into broader, discrete bands, leading to noticeable false contouring artifacts.

## Task 2: Image Operations
Performing subtraction, constant addition, set difference, symmetric difference, and intersection on gray-scale images.

```python
# Fetch standard images from skimage.data
cam = data.camera()
ast = color.rgb2gray(data.astronaut()) * 255
ast = ast.astype(np.uint8)

# Convert to PIL Images and resize
img1 = Image.fromarray(cam)
img2 = Image.fromarray(ast)

resize = (400, 400)
img1 = img1.resize(resize, Image.Resampling.LANCZOS)
img2 = img2.resize(resize, Image.Resampling.LANCZOS)

im1arr = np.asarray(img1, dtype=np.int16)
im2arr = np.asarray(img2, dtype=np.int16)
```

```python
# 1. Subtraction: Image 1 - Image 2
subtraction = np.clip(im1arr - im2arr, 0, 255).astype(np.uint8)
subtraction_img = Image.fromarray(subtraction)

# Display Subtraction Result
plt.figure()
plt.imshow(subtraction_img, cmap='gray')
plt.title('Image Subtraction (Cameraman - Astronaut)')
plt.axis('off')
plt.show()
```

```python
# 2. Addition with a constant value (175)
added_constant = np.clip(im1arr + 175, 0, 255).astype(np.uint8)
added_constant_img = Image.fromarray(added_constant)

# Display Result
plt.figure()
plt.imshow(added_constant_img, cmap='gray')
plt.title('Image + Constant (175)')
plt.axis('off')
plt.show()
```

```python
# Cast arrays to uint8 for logical/bitwise operations
im1_u8 = im1arr.astype(np.uint8)
im2_u8 = im2arr.astype(np.uint8)

# 3. Set Difference: A - B = A AND (NOT B)
set_diff = cv2.bitwise_and(im1_u8, cv2.bitwise_not(im2_u8))

# 4. Symmetric Difference: (A - B) UNION (B - A) or Bitwise XOR
sym_diff = cv2.bitwise_xor(im1_u8, im2_u8)

# 5. Intersection: A AND B
intersection = cv2.bitwise_and(im1_u8, im2_u8)
```

```python
# Plot Set and Logical Operations
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(set_diff, cmap='gray')
axes[0].set_title('Set Difference (A AND NOT B)')
axes[0].axis('off')

axes[1].imshow(sym_diff, cmap='gray')
axes[1].set_title('Symmetric Difference (A XOR B)')
axes[1].axis('off')

axes[2].imshow(intersection, cmap='gray')
axes[2].set_title('Intersection (A AND B)')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
